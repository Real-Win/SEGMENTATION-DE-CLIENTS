## PROJET DE CLUSTERING : SEGMENTATION DE CLIENTS

```python
import numpy as np
import pandas as pd
import seaborn as sns
import plotly.express as px
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
```

```python
url = "https://raw.githubusercontent.com/PacktPublishing/Data-Science-for-Marketing-Analytics-Second-Edition/master/Chapter03/Datasets/Mall_Customers.csv"
df = pd.read_csv(url)
print(df.shape)   
df.head()
```

```python
df.rename(columns={'Gender': 'Gender-Bin'}, inplace=True)
df['Gender-Bin'] = (df['Gender-Bin'] == 'Male').astype(int)
```

```python
df.head()
```

```python
df=df.drop('CustomerID',axis=1)
```

```python
scaler=StandardScaler()
df_scale=scaler.fit_transform(df)
```

```python
inertias=[]
K_range=range(2,11)
silhouette_scores=[]
X = df_scale[:, [2, 3]]
for k in K_range :
    kmeans=KMeans(n_init=10,n_clusters=k,random_state=42)
    labels=kmeans.fit_predict(X)
    score = silhouette_score(X, labels)
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(score)
    print(f"Pour k = {k}, Score de Silhouette = {score:.4f}")

k_optimal = K_range[silhouette_scores.index(max(silhouette_scores))]
print(f"\n---> Le k optimal calculé est : k = {k_optimal}")

# Tracé du graphique du coude
plt.figure(figsize=(8, 4))
plt.plot(K_range, inertias, 'bo-')
plt.xlabel('Nombre de clusters (k)')
plt.ylabel('Inertie')
plt.title('Méthode du Coude pour trouver k optimal')
plt.grid(True)
plt.show()
```

```python
kmeans_final = KMeans(n_init=10,n_clusters=5, random_state=42)
df['Segment_Final'] = kmeans_final.fit_predict(X)

#  Profil complet de chaque segment (y compris Age et Genre !)
profil_final = df.groupby('Segment_Final').agg({
    'Age': 'mean',
    'Gender-Bin': 'mean', # Donnera la proportion d'hommes
    'Annual Income (k$)': 'mean',
    'Spending Score (1-100)': 'mean',
    'Segment_Final': 'count'
}).rename(columns={'Segment_Final': 'Effectif'}).round(2)

#  Affichage du tableau d'analyse
profil_final
```

```python
# Noms métiers clairs
noms_segments = {
    0: 'Clientèle moyenne',
    1: 'Revenus élevés, forte dépense (VIP)',
    2: 'Jeunes, petit budget, forte dépense',
    3: 'Revenus élevés, faible dépense',
    4: 'Petit budget, faible dépense'
}

df['Nom_Segment'] = df['Segment_Final'].map(noms_segments)
df[['Age', 'Annual Income (k$)', 'Spending Score (1-100)', 'Nom_Segment']].head(10)
```

```python

# Création du graphique 3D interactif avec Plotly
fig = px.scatter_3d(
    df, 
    x='Age', 
    y='Annual Income (k$)', 
    z='Spending Score (1-100)',
    color='Nom_Segment', # Couleur par segment
    color_discrete_sequence=px.colors.qualitative.Set1, # Palette de couleurs
    opacity=0.8,
    title='Représentation 3D des Segments Clients'
)

# Personnalisation de la taille et affichage
fig.update_traces(marker=dict(size=6, line=dict(width=1, color='Black')))
fig.update_layout(margin=dict(l=0, r=0, b=0, t=40))

# Affichage dans le notebook
fig.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# On définit une palette de couleurs distinctes
palette = ['#e41a1c', '#377eb8', '#4daf4a', '#984ea3', '#ff7f00']

plt.figure(figsize=(10, 7))

# Création du scatter plot 2D
sns.scatterplot(
    data=df, 
    x='Annual Income (k$)', 
    y='Spending Score (1-100)', 
    hue='Nom_Segment', # On utilise les noms clairs pour la légende
    palette=palette, 
    s=150, # Taille des points
    alpha=0.8, # Transparence
    edgecolor='black' # Bordure des points
)

# Personnalisation des axes et du titre
plt.title('Répartition des Segments Clients (Revenu vs Score de Dépense)', fontsize=16, fontweight='bold')
plt.xlabel('Revenu Annuel (k$)', fontsize=14)
plt.ylabel('Score de Dépense (1-100)', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.5)

# Placement de la légende en dehors du graphique
plt.legend(title='Segment Client', bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=12)

plt.show()
```
